# 2 Automatic differentiation

```python
import torch
```

By default, a tensor doesn't require gradients to be taken with respect zo itself. 

```python
x = torch.ones((3,))
print(x)
print(x.requires_grad)
```

We can tell Pytorch, that we want to take derivatives w.r.t. x

```python
x.requires_grad_(True)
print(x.requires_grad)
```

Lets illustrate AD for an example

```python
x = torch.arange(3, dtype=torch.float32, requires_grad=True) # Only float tensors can have gradients
print("X", x)
```

```python
a = x + 2
b = a ** 2
c = b + 3
y = c.mean()
print("Y", y)
```

Now we compute the derivative of $y$ with respect to any tensor that we tagged by 'requires_grad = True'

```python
y.backward()
```

The gradients is now in 

```python
print(x.grad)
```

```python
x.grad.zero_()  # Reset the gradient
#x = torch.arange(3, dtype=torch.float32, requires_grad=True) # Only float tensors can have gradients
z = torch.norm(torch.sigmoid(x) + x)
print(z)
z.backward()
print(x.grad)
```

### GPUs and Devices

Check, whether there is a GPU

```python
gpu_avail = torch.cuda.is_available()
print(f"Is the GPU available? {gpu_avail}")
```

Get the device

```python
device = torch.device("cuda") if torch.cuda.is_available() else torch.device("cpu")
print("Device", device)
```

Push a tensor to the GPU

```python
x = torch.zeros(2, 3)
x = x.to(device)
print("X", x)
```
